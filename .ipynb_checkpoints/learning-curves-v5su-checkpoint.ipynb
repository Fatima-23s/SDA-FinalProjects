```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

RUN_NAME = "v5su_640_newsplit"

# Search the usual places for this run's results.csv
SEARCH_ROOTS = [Path("/kaggle/input"), Path("/kaggle/working")]
matches = []
for root in SEARCH_ROOTS:
    if root.exists():
        matches += list(root.rglob(f"{RUN_NAME}*/results.csv"))
print("Found:", matches)

if not matches:
    raise FileNotFoundError(f"results.csv for {RUN_NAME} not found. Set RESULTS_CSV manually.")
RESULTS_CSV = matches[0]

history = pd.read_csv(RESULTS_CSV)
history.columns = history.columns.str.strip()

required = [
    "epoch",
    "train/box_loss", "val/box_loss",
    "train/cls_loss", "val/cls_loss",
]
missing = [c for c in required if c not in history.columns]
if missing:
    raise ValueError(f"Missing columns: {missing}")

fig, axes = plt.subplots(1, 2, figsize=(13, 5))
for ax, loss, title in zip(
    axes,
    ["box", "cls"],
    ["Bounding Box Loss", "Classification Loss"],
):
    ax.plot(history["epoch"], history[f"train/{loss}_loss"],
            label="Training", color="#4C72B0", linewidth=1.8)
    ax.plot(history["epoch"], history[f"val/{loss}_loss"],
            label="Validation", color="#DD8452", linewidth=1.8)
    ax.set_title(title)
    ax.set_xlabel("Epoch")
    ax.set_ylabel("Loss")
    ax.grid(alpha=0.3)
    ax.legend()

fig.suptitle("Learning Curves — Re-split Dataset (YOLOv5su 640)")
plt.tight_layout()
plt.savefig("learning_curves_v5su.png", dpi=200, bbox_inches="tight")
plt.show()

print(f"Training log: {RESULTS_CSV}")
print(f"Last recorded epoch: {int(history['epoch'].iloc[-1])}")
print("Figure saved to: learning_curves_v5su.png")
```
